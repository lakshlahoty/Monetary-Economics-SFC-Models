```python
def describe(v):
    #Turns a qualitative value into a plain-English word.
    if v == "up":
        return "increased"
    elif v == "down":
        return "decreased"
    elif v == "same":
        return "stayed the same"
    elif v == "unknown":
        return "is unknown"
    else:
        return v


def ask_user(a, b, a_descr, b_descr, outcome, context, period):
    #Called whenever the sign pattern does not give a unique answer.
    print()
    print(f"Period {period} \u2014 {outcome}")
    print("-" * 40)
    print(f"{a_descr} {describe(a)} while {b_descr} {describe(b)} ({context}).")
    print("The result cannot be determined from the signs alone.")
    print(f"\nShould {outcome} be:")
    print("  up")
    print("  down")
    print("  same")
    print("  unknown")
    choice = input("\nEnter your choice: ").strip().lower()
    while choice not in ("up", "down", "same", "unknown"):
        print("Please type one of: up, down, same, unknown")
        choice = input("Enter your choice: ").strip().lower()
    return choice


def make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous):
    #Builds the plain-English sentence describing how the result was found.
    if ambiguous:
        return (f"{a_descr} {describe(a)} while {b_descr} {describe(b)}.\n"
                f"The result for {outcome} cannot be determined uniquely.\n"
                f"User selected: {result}.")
    else:
        return (f"{a_descr} {describe(a)} and {b_descr} {describe(b)}, "
                f"therefore {outcome} {describe(result)}.")


def add(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "up") | ("up", "same") | ("same", "up"):
            result = "up"
        case ("down", "down") | ("down", "same") | ("same", "down"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "down") | ("down", "up") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def sub(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "down") | ("up", "same") | ("same", "down"):
            result = "up"
        case ("down", "up") | ("down", "same") | ("same", "up"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "up") | ("down", "down") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def mult(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "up") | ("up", "same") | ("same", "up"):
            result = "up"
        case ("down", "down") | ("down", "same") | ("same", "down"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "down") | ("down", "up") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def div(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "down") | ("up", "same") | ("same", "down"):
            result = "up"
        case ("down", "up") | ("down", "same") | ("same", "up"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "up") | ("down", "down") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def identity(a, a_descr, outcome, context, period=None):
    #Used for pure definitional equalities (e.g. r_m = r_l = r_bar).
    #Copying a single qualitative value is never ambiguous.
    result = a
    expl = (f"By definition, {outcome} equals {a_descr}, which {describe(a)} "
            f"({context}).")
    return (result, expl)


class QualitativeEconomyBMW:
    def __init__(self, K_1="same", L_1="same", Mh_1="same", Y_1="same", r_1="same"):
        self.period = [0]

        #Exogenous / policy inputs, set fresh every period.
        self.alpha0 = ["same"]
        self.alpha1 = ["same"]
        self.alpha2 = ["same"]
        self.delta = ["same"]
        self.gamma = ["same"]
        self.kappa = ["same"]
        self.pr = ["same"]
        self.r_bar = ["same"]

        #Computed flows.
        self.AF = ["same"]
        self.KT = ["same"]
        self.I = ["same"]
        self.WB = ["same"]
        self.YD = ["same"]
        self.C = ["same"]
        self.Y = [Y_1]
        self.N = ["same"]
        self.W = ["same"]
        self.dL = ["same"]
        self.dMh = ["same"]
        self.dK = ["same"]

        #Stocks (levels carried across periods).
        self.K = [K_1]
        self.L = [L_1]
        self.Mh = [Mh_1]
        self.r = [r_1]

        #One explanation list per calculated variable, so a full
        #reasoning trace can be printed later with printExplanations().
        self.AF_expl = [""]
        self.KT_expl = [""]
        self.I_expl = [""]
        self.WB_expl = [""]
        self.YD_expl = [""]
        self.C_expl = [""]
        self.Y_expl = [""]
        self.N_expl = [""]
        self.W_expl = [""]
        self.dL_expl = [""]
        self.dMh_expl = [""]
        self.dK_expl = [""]
        self.K_expl = [""]
        self.L_expl = [""]
        self.Mh_expl = [""]
        self.r_expl = [""]

    def get_K(self):
        return self.K[-1]

    def get_L(self):
        return self.L[-1]

    def get_Mh(self):
        return self.Mh[-1]

    def get_Y(self):
        return self.Y[-1]

    def get_r(self):
        return self.r[-1]

    def set_alpha0(self, v):
        self.alpha0.append(v)

    def set_alpha1(self, v):
        self.alpha1.append(v)

    def set_alpha2(self, v):
        self.alpha2.append(v)

    def set_delta(self, v):
        self.delta.append(v)

    def set_gamma(self, v):
        self.gamma.append(v)

    def set_kappa(self, v):
        self.kappa.append(v)

    def set_pr(self, v):
        self.pr.append(v)

    def set_r_bar(self, v):
        self.r_bar.append(v)

    def set_AF(self, v):
        self.AF.append(v)

    def set_KT(self, v):
        self.KT.append(v)

    def set_I(self, v):
        self.I.append(v)

    def set_WB(self, v):
        self.WB.append(v)

    def set_YD(self, v):
        self.YD.append(v)

    def set_C(self, v):
        self.C.append(v)

    def set_Y(self, v):
        self.Y.append(v)

    def set_N(self, v):
        self.N.append(v)

    def set_W(self, v):
        self.W.append(v)

    def set_dL(self, v):
        self.dL.append(v)

    def set_dMh(self, v):
        self.dMh.append(v)

    def set_dK(self, v):
        self.dK.append(v)

    def set_K(self, v):
        self.K.append(v)

    def set_L(self, v):
        self.L.append(v)

    def set_Mh(self, v):
        self.Mh.append(v)

    def set_r(self, v):
        self.r.append(v)

    def progresCycle(self, alpha0, alpha1, alpha2, delta, gamma, kappa, pr, r_bar):
        #The period we are about to compute (used only to label prompts and the explanation trace - it does not affect the logic).
        period_number = self.period[-1] + 1

        K_1 = self.get_K()
        L_1 = self.get_L()
        Mh_1 = self.get_Mh()
        Y_1 = self.get_Y()
        r_1 = self.get_r()

        #Firms: amortization funds / depreciation allowance
        AF, AF_expl = mult(delta, K_1,
                            "The depreciation rate (delta)",
                            "The previous capital stock",
                            "Amortization funds (AF), which also equals the "
                            "depreciation allowance (DA)",
                            "AF = DA = delta * K_{-1}  (eq. 7.7 / 7.18)",
                            period_number)

        #Firms: investment
        KT, KT_expl = mult(kappa, Y_1,
                            "The target capital-to-output ratio (kappa)",
                            "Income in the previous period",
                            "The target capital stock (K^T)",
                            "K^T = kappa * Y_{-1}  (eq. 7.19)",
                            period_number)

        gap, gap_expl = sub(KT, K_1,
                             "The target capital stock",
                             "The previous capital stock",
                             "The capital-stock adjustment gap",
                             "gap = K^T - K_{-1}  (part of eq. 7.20)",
                             period_number)

        adj, adj_expl = mult(gamma, gap,
                              "The investment adjustment speed (gamma)",
                              "The capital-stock adjustment gap",
                              "The stock-adjustment component of investment",
                              "adj = gamma * (K^T - K_{-1})  (part of eq. 7.20)",
                              period_number)

        I, I_expl_add = add(adj, AF,
                             "The stock-adjustment component of investment",
                             "Amortization funds / depreciation allowance",
                             "Investment (I), where I_s = I_d",
                             "I = gamma*(K^T - K_{-1}) + DA  (eq. 7.20, with eq. 7.2)",
                             period_number)
        I_expl = KT_expl + "\n" + gap_expl + "\n" + adj_expl + "\n" + I_expl_add

        #Firms: wage bill 
        Y_e = Y_1  # expected income = last period's actual income

        interest_L, interest_L_expl = mult(r_1, L_1,
                                            "The (lagged) interest rate",
                                            "The previous stock of loans",
                                            "Interest paid on loans",
                                            "r_{-1} * L_{-1}  (part of eq. 7.6)",
                                            period_number)

        WB_partial, WB_partial_expl = sub(Y_e, interest_L,
                                           "Expected income (last period's actual income)",
                                           "Interest paid on loans",
                                           "The wage bill before amortization",
                                           "Y_{-1} - r_{-1}*L_{-1}  (part of eq. 7.6)",
                                           period_number)

        WB, WB_expl_add = sub(WB_partial, AF,
                               "The wage bill before amortization",
                               "Amortization funds",
                               "The wage bill (WB), where WB_s = WB_d",
                               "WB = Y_{-1} - r_{-1}*L_{-1} - AF  "
                               "(eq. 7.6, using expected income Y_e = Y_{-1})",
                               period_number)
        WB_expl = interest_L_expl + "\n" + WB_partial_expl + "\n" + WB_expl_add

        #Households: disposable income and consumption
        interest_M, interest_M_expl = mult(r_1, Mh_1,
                                            "The (lagged) interest rate",
                                            "The previous stock of household money",
                                            "Interest received on deposits",
                                            "r_{-1} * M_{h,-1}  (part of eq. 7.9)",
                                            period_number)

        YD, YD_expl_add = add(WB, interest_M,
                               "The wage bill",
                               "Interest received on deposits",
                               "Disposable income (YD)",
                               "YD = WB + r_{-1}*M_{h,-1}  (eq. 7.9)",
                               period_number)
        YD_expl = interest_M_expl + "\n" + YD_expl_add

        term_YD, term_YD_expl = mult(alpha1, YD,
                                      "The propensity to consume out of income (alpha1)",
                                      "Disposable income",
                                      "The income effect on consumption",
                                      "alpha1 * YD  (part of eq. 7.16)",
                                      period_number)

        term_H, term_H_expl = mult(alpha2, Mh_1,
                                    "The propensity to consume out of wealth (alpha2)",
                                    "Household wealth in the previous period",
                                    "The wealth effect on consumption",
                                    "alpha2 * M_{h,-1}  (part of eq. 7.16)",
                                    period_number)

        C_partial, C_partial_expl = add(alpha0, term_YD,
                                         "Autonomous consumption (alpha0)",
                                         "The income effect on consumption",
                                         "Consumption before the wealth effect",
                                         "alpha0 + alpha1*YD  (part of eq. 7.16)",
                                         period_number)

        C, C_expl_add = add(C_partial, term_H,
                             "Consumption before the wealth effect",
                             "The wealth effect on consumption",
                             "Consumption (C), where C_s = C_d",
                             "C = alpha0 + alpha1*YD + alpha2*M_{h,-1}  (eq. 7.16)",
                             period_number)
        C_expl = (term_YD_expl + "\n" + term_H_expl + "\n"
                  + C_partial_expl + "\n" + C_expl_add)

        #Income
        Y, Y_expl = add(C, I,
                         "Consumption", "Investment",
                         "Income (Y)",
                         "Y = C + I  (eq. 7.5)",
                         period_number)

        #Employment and the wage rate (use ACTUAL current Y: no circularity from here on)
        N, N_expl = div(Y, pr,
                         "Income", "Labour productivity (pr)",
                         "Employment (N), where N_s = N_d",
                         "N = Y / pr  (eq. 7.14)",
                         period_number)

        W, W_expl = div(WB, N,
                         "The wage bill", "Employment",
                         "The wage rate (W)",
                         "W = WB / N  (eq. 7.15)",
                         period_number)

        #Flows of funds
        dL, dL_expl = sub(I, AF,
                           "Investment", "Amortization funds",
                           "The change in loans / bank deposits "
                           "(delta L = delta M_s)",
                           "delta L = I - AF  (eq. 7.8, with eq. 7.4 and 7.11)",
                           period_number)

        dMh, dMh_expl = sub(YD, C,
                             "Disposable income", "Consumption",
                             "The change in household money holdings (delta M_h)",
                             "delta M_h = YD - C  (eq. 7.10)",
                             period_number)

        dK, dK_expl = sub(I, AF,
                           "Investment", "Amortization funds / depreciation allowance",
                           "The change in the capital stock (delta K)",
                           "delta K = I - DA  (eq. 7.17)",
                           period_number)

        #Stock updates
        K, K_expl_add = add(K_1, dK,
                             "The previous capital stock",
                             "The change in the capital stock",
                             "The capital stock (K)",
                             "K = K_{-1} + delta K",
                             period_number)
        K_expl = dK_expl + "\n" + K_expl_add

        L, L_expl_add = add(L_1, dL,
                             "The previous stock of loans",
                             "The change in loans",
                             "The stock of loans (L)",
                             "L = L_{-1} + delta L",
                             period_number)
        L_expl = dL_expl + "\n" + L_expl_add

        Mh, Mh_expl_add = add(Mh_1, dMh,
                               "The previous stock of household money",
                               "The change in household money holdings",
                               "The stock of household money (M_h)",
                               "M_h = M_{h,-1} + delta M_h",
                               period_number)
        Mh_expl = dMh_expl + "\n" + Mh_expl_add

        r, r_expl = identity(r_bar,
                              "The exogenous policy rate (r_bar)",
                              "Both the deposit rate and the loan rate (r)",
                              "r_m = r_l = r_bar  (eq. 7.12 and eq. 7.21)",
                              period_number)

        self.period.append(period_number)
        self.set_alpha0(alpha0)
        self.set_alpha1(alpha1)
        self.set_alpha2(alpha2)
        self.set_delta(delta)
        self.set_gamma(gamma)
        self.set_kappa(kappa)
        self.set_pr(pr)
        self.set_r_bar(r_bar)

        self.set_AF(AF)
        self.set_KT(KT)
        self.set_I(I)
        self.set_WB(WB)
        self.set_YD(YD)
        self.set_C(C)
        self.set_Y(Y)
        self.set_N(N)
        self.set_W(W)
        self.set_dL(dL)
        self.set_dMh(dMh)
        self.set_dK(dK)
        self.set_K(K)
        self.set_L(L)
        self.set_Mh(Mh)
        self.set_r(r)

        self.AF_expl.append(AF_expl)
        self.KT_expl.append(KT_expl)
        self.I_expl.append(I_expl)
        self.WB_expl.append(WB_expl)
        self.YD_expl.append(YD_expl)
        self.C_expl.append(C_expl)
        self.Y_expl.append(Y_expl)
        self.N_expl.append(N_expl)
        self.W_expl.append(W_expl)
        self.dL_expl.append(dL_expl)
        self.dMh_expl.append(dMh_expl)
        self.dK_expl.append(dK_expl)
        self.K_expl.append(K_expl)
        self.L_expl.append(L_expl)
        self.Mh_expl.append(Mh_expl)
        self.r_expl.append(r_expl)

    def printHistory(self):
        #The model has more variables than Chapter 3 but usually only a
        #handful of periods, so variables run down the rows and periods
        #run across the columns, exactly as in the Chapter 3 table.
        variables = [
            ("alpha0", self.alpha0), ("alpha1", self.alpha1),
            ("alpha2", self.alpha2), ("delta", self.delta),
            ("gamma", self.gamma), ("kappa", self.kappa),
            ("pr", self.pr), ("r_bar", self.r_bar),
            ("AF/DA", self.AF), ("K^T", self.KT), ("I", self.I),
            ("WB", self.WB), ("YD", self.YD), ("C", self.C), ("Y", self.Y),
            ("N", self.N), ("W", self.W), ("dL", self.dL),
            ("dMh", self.dMh), ("dK", self.dK),
            ("K", self.K), ("L", self.L), ("Mh", self.Mh), ("r", self.r),
        ]

        label_width = 8
        col_width = 10

        header_line = f"{'Variable':<{label_width}}" + "".join(
            f"{'Period ' + str(p):>{col_width}}" for p in self.period)
        total_width = len(header_line)

        print("=" * total_width)
        print("SIMULATION RESULTS \u2014 Model BMW")
        print("=" * total_width)
        print(header_line)
        print("-" * total_width)

        for name, values in variables:
            row_line = f"{name:<{label_width}}" + "".join(
                f"{str(v):>{col_width}}" for v in values)
            print(row_line)

        print("=" * total_width)

        #Ratio of unknown / same / up / down answers across the table 
        #Counts every value in every variable's history, excluding each
        #variable's period-0 initial condition (that is a starting point,
        #not a computed answer). The numbers are unchanged from before 
        #only how they're displayed (via _print_answer_distribution) has
        #changed.
        counts = {"up": 0, "down": 0, "same": 0, "unknown": 0}
        for name, values in variables:
            for v in values[1:]:
                if v in counts:
                    counts[v] += 1
        total = sum(counts.values())
        if total > 0:
            self._print_answer_distribution(counts, total)

    def _print_answer_distribution(self, counts, total):
        #A small, purely cosmetic "dashboard" box summarising how the
        #up/down/same/unknown answers were distributed across every
        #computed value in the table above. The counts themselves come
        #straight from printHistory() - this function only formats them.
        box_width = 64
        inner = box_width - 2

        def boxed(text=""):
            return "\u2502" + text.ljust(inner) + "\u2502"

        top = "\u256d" + "\u2500" * inner + "\u256e"
        divider = "\u251c" + "\u2500" * inner + "\u2524"
        bottom = "\u2570" + "\u2500" * inner + "\u256f"

        print()
        print(top)
        print(boxed(" ANSWER DISTRIBUTION".center(inner)))
        print(divider)
        print(boxed(f"  Unknown : Same : Up : Down  =  {counts['unknown']} : "
                     f"{counts['same']} : {counts['up']} : {counts['down']}"
                     f"   (n = {total})"))
        print(boxed())

        bar_slots = 24
        for key in ("unknown", "same", "up", "down"):
            pct = counts[key] / total
            filled = round(pct * bar_slots)
            bar = "\u2588" * filled + "\u2591" * (bar_slots - filled)
            row = f"  {key.capitalize():<8}{bar}  {pct:>5.1%}"
            print(boxed(row))

        print(bottom)

    def printExplanations(self):
        #Prints the full plain-English reasoning trace for every period.
        #Each variable's heading is underlined and its explanation lines
        #are shown as bullet points - purely a display change, the
        #underlying text is exactly what was generated in progresCycle().
        width = 60
        for i in range(1, len(self.period)):
            print()
            print("=" * width)
            print(f"PERIOD {self.period[i]} \u2014 REASONING TRACE".center(width))
            print("=" * width)

            sections = [
                ("Amortization funds / depreciation allowance (AF = DA)", self.AF_expl[i]),
                ("Investment (I)", self.I_expl[i]),
                ("Wage bill (WB)", self.WB_expl[i]),
                ("Disposable income (YD)", self.YD_expl[i]),
                ("Consumption (C)", self.C_expl[i]),
                ("Income (Y)", self.Y_expl[i]),
                ("Employment (N)", self.N_expl[i]),
                ("Wage rate (W)", self.W_expl[i]),
                ("Change in loans / bank money (delta L)", self.dL_expl[i]),
                ("Change in household money holdings (delta M_h)", self.dMh_expl[i]),
                ("Change in the capital stock (delta K)", self.dK_expl[i]),
                ("Capital stock (K)", self.K_expl[i]),
                ("Stock of loans (L)", self.L_expl[i]),
                ("Stock of household money (M_h)", self.Mh_expl[i]),
                ("Interest rate (r)", self.r_expl[i]),
            ]

            for title, text in sections:
                print()
                print(title)
                print("\u2500" * len(title))
                for line_text in text.split("\n"):
                    if line_text.strip():
                        print(f"  \u2022 {line_text}")

        print()
        print("=" * width)


qecon = QualitativeEconomyBMW(K_1="same", L_1="same", Mh_1="same",
                               Y_1="same", r_1="same")
#Period 1: a rise in autonomous consumption (alpha0 up), everything else same.
qecon.progresCycle(alpha0="up", alpha1="same", alpha2="same", delta="same",
                    gamma="same", kappa="same", pr="same", r_bar="same")
#Period 2: the central bank raises its policy rate (r_bar up), everything else same.
qecon.progresCycle(alpha0="same", alpha1="same", alpha2="same", delta="same",
                    gamma="same", kappa="same", pr="same", r_bar="up")
qecon.printHistory()
qecon.printExplanations()
```
