```python
def describe(v):
    #Turns a qualitative value into a plain-English word.
    if v == "up":
        return "increased"
    elif v == "down":
        return "decreased"
    elif v == "same":
        return "stayed the same"
    elif v == "unknown":
        return "is unknown"
    else:
        return v


def ask_user(a, b, a_descr, b_descr, outcome, context, period):
    #Called whenever the sign pattern does not give a unique answer.
    print()
    print(f"Period {period} \u2014 {outcome}")
    print("-" * 40)
    print(f"{a_descr} {describe(a)} while {b_descr} {describe(b)} ({context}).")
    print("The result cannot be determined from the signs alone.")
    print(f"\nShould {outcome} be:")
    print("  up")
    print("  down")
    print("  same")
    print("  unknown")
    choice = input("\nEnter your choice: ").strip().lower()
    while choice not in ("up", "down", "same", "unknown"):
        print("Please type one of: up, down, same, unknown")
        choice = input("Enter your choice: ").strip().lower()
    return choice


def make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous):
    #Builds the plain-English sentence describing how the result was found.
    if ambiguous:
        return (f"{a_descr} {describe(a)} while {b_descr} {describe(b)}.\n"
                f"The result for {outcome} cannot be determined uniquely.\n"
                f"User selected: {result}.")
    else:
        return (f"{a_descr} {describe(a)} and {b_descr} {describe(b)}, "
                f"therefore {outcome} {describe(result)}.")


def add(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "up") | ("up", "same") | ("same", "up"):
            result = "up"
        case ("down", "down") | ("down", "same") | ("same", "down"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "down") | ("down", "up") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def sub(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "down") | ("up", "same") | ("same", "down"):
            result = "up"
        case ("down", "up") | ("down", "same") | ("same", "up"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "up") | ("down", "down") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def mult(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "up") | ("up", "same") | ("same", "up"):
            result = "up"
        case ("down", "down") | ("down", "same") | ("same", "down"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "down") | ("down", "up") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def div(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "down") | ("up", "same") | ("same", "down"):
            result = "up"
        case ("down", "up") | ("down", "same") | ("same", "up"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "up") | ("down", "down") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def exp(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "up") | ("up", "same") | ("same", "up"):
            result = "up"
        case ("down", "down") | ("down", "same") | ("same", "down"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "down") | ("down", "up") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


class QualitativeEconomy:
    def __init__(self, YD_1="same", H_1="same"):
        self.period = [0]

        #Exogenous / policy inputs, set fresh every period.
        self.G = ["same"]
        self.alpha1 = ["same"]
        self.alpha2 = ["same"]
        self.theta = ["same"]

        #Computed flows.
        self.Y = ["same"]
        self.T = ["same"]
        self.YD = [YD_1]
        self.YD_e = [YD_1]
        self.C = ["same"]
        self.delta_Hs = ["same"]
        self.delta_Hh = ["same"]
        self.delta_Hd = ["same"]
        self.Hd = [H_1]

        #Stocks (levels carried across periods).
        self.H = [H_1]

        #One explanation list per calculated variable, so a full
        #reasoning trace can be printed later with printExplanations().
        self.C_expl = [""]
        self.Y_expl = [""]
        self.T_expl = [""]
        self.YD_expl = [""]
        self.delta_Hs_expl = [""]
        self.delta_Hh_expl = [""]
        self.H_expl = [""]
        self.delta_Hd_expl = [""]
        self.Hd_expl = [""]

    def get_H(self):
        return self.H[-1]

    def get_YD(self):
        return self.YD[-1]

    def set_G(self, v):
        self.G.append(v)

    def set_alpha1(self, v):
        self.alpha1.append(v)

    def set_alpha2(self, v):
        self.alpha2.append(v)

    def set_theta(self, v):
        self.theta.append(v)

    def set_C(self, v):
        self.C.append(v)

    def set_Y(self, v):
        self.Y.append(v)

    def set_T(self, v):
        self.T.append(v)

    def set_YD(self, v):
        self.YD.append(v)

    def set_YD_e(self, v):
        self.YD_e.append(v)

    def set_delta_Hs(self, v):
        self.delta_Hs.append(v)

    def set_delta_Hh(self, v):
        self.delta_Hh.append(v)

    def set_H(self, v):
        self.H.append(v)

    def set_delta_Hd(self, v):
        self.delta_Hd.append(v)

    def set_Hd(self, v):
        self.Hd.append(v)

    def progresCycle(self, G, alpha1, alpha2, theta):
        #The period we are about to compute (used only to label prompts and the explanation trace - it does not affect the logic).
        period_number = self.period[-1] + 1

        H_1 = self.get_H()
        YD_e = self.get_YD()

        #Consumption: C = alpha1 * YD_e + alpha2 * H_1
        term1, term1_expl = mult(alpha1, YD_e,
                                  "The propensity to consume out of income (alpha1)",
                                  "Expected disposable income",
                                  "the income effect on consumption",
                                  "term1 = alpha1 * YD_e", period_number)
        term2, term2_expl = mult(alpha2, H_1,
                                  "The propensity to consume out of wealth (alpha2)",
                                  "Wealth",
                                  "the wealth effect on consumption",
                                  "term2 = alpha2 * H_1", period_number)
        C, C_expl_add = add(term1, term2,
                             "The income effect on consumption",
                             "The wealth effect on consumption",
                             "Consumption (C)",
                             "C = term1 + term2", period_number)
        C_expl = term1_expl + "\n" + term2_expl + "\n" + C_expl_add

        #Income: Y = C + G
        Y, Y_expl = add(C, G, "Consumption", "Government spending",
                         "Income (Y)", "Y = C + G", period_number)

        #Taxes: T = theta * Y
        T, T_expl = mult(theta, Y, "The tax rate (theta)", "Income",
                          "Taxes (T)", "T = theta * Y", period_number)

        #Disposable income: YD = Y - T
        YD, YD_expl = sub(Y, T, "Income", "Taxes",
                           "Disposable income (YD)", "YD = Y - T", period_number)

        #Change in money stock supplied: delta_Hs = G - T
        delta_Hs, delta_Hs_expl = sub(G, T, "Government spending", "Taxes",
                                       "The change in the money supply (delta_Hs)",
                                       "delta_Hs = G - T", period_number)

        #Change in household wealth actually held: delta_Hh = YD - C
        delta_Hh, delta_Hh_expl = sub(YD, C, "Disposable income", "Consumption",
                                       "The change in household wealth (delta_Hh)",
                                       "delta_Hh = YD - C", period_number)

        #Wealth: H = H_1 + delta_Hs
        H, H_expl = add(H_1, delta_Hs, "Previous wealth",
                         "The change in the money supply",
                         "Wealth (H)", "H = H_1 + delta_Hs", period_number)

        #Change in desired wealth: delta_Hd = YD_e - C
        delta_Hd, delta_Hd_expl = sub(YD_e, C, "Expected disposable income", "Consumption",
                                       "The change in desired wealth (delta_Hd)",
                                       "delta_Hd = YD_e - C", period_number)

        #Desired wealth: Hd = H_1 + delta_Hd
        Hd, Hd_expl = add(H_1, delta_Hd, "Previous wealth",
                           "The change in desired wealth",
                           "Desired wealth (Hd)", "Hd = H_1 + delta_Hd", period_number)

        self.period.append(period_number)
        self.set_G(G)
        self.set_alpha1(alpha1)
        self.set_alpha2(alpha2)
        self.set_theta(theta)
        self.set_C(C)
        self.set_Y(Y)
        self.set_T(T)
        self.set_YD(YD)
        self.set_YD_e(YD_e)
        self.set_delta_Hs(delta_Hs)
        self.set_delta_Hh(delta_Hh)
        self.set_H(H)
        self.set_delta_Hd(delta_Hd)
        self.set_Hd(Hd)

        self.C_expl.append(C_expl)
        self.Y_expl.append(Y_expl)
        self.T_expl.append(T_expl)
        self.YD_expl.append(YD_expl)
        self.delta_Hs_expl.append(delta_Hs_expl)
        self.delta_Hh_expl.append(delta_Hh_expl)
        self.H_expl.append(H_expl)
        self.delta_Hd_expl.append(delta_Hd_expl)
        self.Hd_expl.append(Hd_expl)

    def printHistory(self):
        #The model has 14 variables but usually only a handful of periods,
        #so listing variables down the rows and periods across the columns
        #keeps the table narrow and readable in any terminal width.
        variables = [
            ("G", self.G), ("a1", self.alpha1), ("a2", self.alpha2),
            ("theta", self.theta), ("Y", self.Y), ("T", self.T),
            ("YD", self.YD), ("YD^e", self.YD_e), ("C", self.C),
            ("dHs", self.delta_Hs), ("dHh", self.delta_Hh), ("H", self.H),
            ("dHd", self.delta_Hd), ("Hd", self.Hd),
        ]

        label_width = 8
        col_width = 10

        header_line = f"{'Variable':<{label_width}}" + "".join(
            f"{'Period ' + str(p):>{col_width}}" for p in self.period)
        total_width = len(header_line)

        print("=" * total_width)
        print("SIMULATION RESULTS \u2014 Model SIM")
        print("=" * total_width)
        print(header_line)
        print("-" * total_width)

        for name, values in variables:
            row_line = f"{name:<{label_width}}" + "".join(
                f"{str(v):>{col_width}}" for v in values)
            print(row_line)

        print("=" * total_width)

        #Ratio of unknown / same / up / down answers across the table
        #Counts every value in every variable's history, excluding each
        #variable's period-0 initial condition (that is a starting point,
        #not a computed answer). The numbers are unchanged from before
        #only how they're displayed has changed.
        counts = {"up": 0, "down": 0, "same": 0, "unknown": 0}
        for name, values in variables:
            for v in values[1:]:
                if v in counts:
                    counts[v] += 1
        total = sum(counts.values())
        if total > 0:
            self._print_answer_distribution(counts, total)

    def _print_answer_distribution(self, counts, total):
        #A small, purely cosmetic "dashboard" box summarising how the
        #up/down/same/unknown answers were distributed across every
        #computed value in the table above. The counts themselves come
        #straight from printHistory() - this function only formats them.
        box_width = 64
        inner = box_width - 2

        def boxed(text=""):
            return "\u2502" + text.ljust(inner) + "\u2502"

        top = "\u256d" + "\u2500" * inner + "\u256e"
        divider = "\u251c" + "\u2500" * inner + "\u2524"
        bottom = "\u2570" + "\u2500" * inner + "\u256f"

        print()
        print(top)
        print(boxed(" ANSWER DISTRIBUTION".center(inner)))
        print(divider)
        print(boxed(f"  Unknown : Same : Up : Down  =  {counts['unknown']} : "
                     f"{counts['same']} : {counts['up']} : {counts['down']}"
                     f"   (n = {total})"))
        print(boxed())

        bar_slots = 24
        for key in ("unknown", "same", "up", "down"):
            pct = counts[key] / total
            filled = round(pct * bar_slots)
            bar = "\u2588" * filled + "\u2591" * (bar_slots - filled)
            row = f"  {key.capitalize():<8}{bar}  {pct:>5.1%}"
            print(boxed(row))

        print(bottom)

    def printExplanations(self):
        #Prints the full plain-English reasoning trace for every period.
        #Each variable's heading is underlined and its explanation lines
        #are shown as bullet points - purely a display change, the
        #underlying text is exactly what was generated in progresCycle().
        width = 60
        for i in range(1, len(self.period)):
            print()
            print("=" * width)
            print(f"PERIOD {self.period[i]} \u2014 REASONING TRACE".center(width))
            print("=" * width)

            sections = [
                ("Consumption (C)", self.C_expl[i]),
                ("Income (Y)", self.Y_expl[i]),
                ("Taxes (T)", self.T_expl[i]),
                ("Disposable income (YD)", self.YD_expl[i]),
                ("Change in money supply (delta_Hs)", self.delta_Hs_expl[i]),
                ("Change in household wealth (delta_Hh)", self.delta_Hh_expl[i]),
                ("Wealth (H)", self.H_expl[i]),
                ("Change in desired wealth (delta_Hd)", self.delta_Hd_expl[i]),
                ("Desired wealth (Hd)", self.Hd_expl[i]),
            ]

            for title, text in sections:
                print()
                print(title)
                print("\u2500" * len(title))
                for line_text in text.split("\n"):
                    if line_text.strip():
                        print(f"  \u2022 {line_text}")

        print()
        print("=" * width)


qecon = QualitativeEconomy(YD_1="same", H_1="same")
#Period 1: a rise in government spending (G up), everything else same.
qecon.progresCycle("up", "same", "same", "same")
#Period 2: no further changes, everything else same.
qecon.progresCycle("same", "same", "same", "same")
qecon.printHistory()
qecon.printExplanations()
```
