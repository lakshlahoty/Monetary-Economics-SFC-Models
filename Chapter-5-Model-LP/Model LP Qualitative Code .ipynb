```python
def describe(v):
    #Turns a qualitative value into a plain-English word.
    if v == "up":
        return "increased"
    elif v == "down":
        return "decreased"
    elif v == "same":
        return "stayed the same"
    elif v == "unknown":
        return "is unknown"
    else:
        return v


def ask_user(a, b, a_descr, b_descr, outcome, context, period):
    #Called whenever the sign pattern does not give a unique answer.
    print()
    print(f"Period {period} \u2014 {outcome}")
    print("-" * 40)
    print(f"{a_descr} {describe(a)} while {b_descr} {describe(b)} ({context}).")
    print("The result cannot be determined from the signs alone.")
    print(f"\nShould {outcome} be:")
    print("  up")
    print("  down")
    print("  same")
    print("  unknown")
    choice = input("\nEnter your choice: ").strip().lower()
    while choice not in ("up", "down", "same", "unknown"):
        print("Please type one of: up, down, same, unknown")
        choice = input("Enter your choice: ").strip().lower()
    return choice


def make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous):
    #Builds the plain-English sentence describing how the result was found.
    if ambiguous:
        return (f"{a_descr} {describe(a)} while {b_descr} {describe(b)}.\n"
                f"The result for {outcome} cannot be determined uniquely.\n"
                f"User selected: {result}.")
    else:
        return (f"{a_descr} {describe(a)} and {b_descr} {describe(b)}, "
                f"therefore {outcome} {describe(result)}.")


def add(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "up") | ("up", "same") | ("same", "up"):
            result = "up"
        case ("down", "down") | ("down", "same") | ("same", "down"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "down") | ("down", "up") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def sub(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "down") | ("up", "same") | ("same", "down"):
            result = "up"
        case ("down", "up") | ("down", "same") | ("same", "up"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "up") | ("down", "down") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def mult(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "up") | ("up", "same") | ("same", "up"):
            result = "up"
        case ("down", "down") | ("down", "same") | ("same", "down"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "down") | ("down", "up") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def div(a, b, a_descr, b_descr, outcome, context, period=None):
    ambiguous = False
    match (a, b):
        case ("up", "down") | ("up", "same") | ("same", "down"):
            result = "up"
        case ("down", "up") | ("down", "same") | ("same", "up"):
            result = "down"
        case ("same", "same"):
            result = "same"
        case ("up", "up") | ("down", "down") | ("unknown", _) | (_, "unknown"):
            ambiguous = True
            result = ask_user(a, b, a_descr, b_descr, outcome, context, period)
        case _:
            return ("Error: case unhandled", "Error: case unhandled")
    expl = make_explanation(a, b, a_descr, b_descr, outcome, result, ambiguous)
    return (result, expl)


def identity(a, a_descr, outcome, context, period=None):
    #Pure definitional equality -- copying a value is never ambiguous.
    result = a
    expl = (f"By definition, {outcome} equals {a_descr}, which {describe(a)} "
            f"({context}).")
    return (result, expl)


class QualitativeEconomyLP:
    def __init__(self, V_1="same", YDr_1="same", Bh_1="same", BLh_1="same",
                 Bs_1="same", Bcb_1="same", Hs_1="same", r_b_1="same", P_bL_1="same"):
        self.period = [0]

        #Exogenous / policy inputs, set fresh every period.
        self.G = ["same"]
        self.alpha1 = ["same"]
        self.alpha2 = ["same"]
        self.theta = ["same"]
        self.lambda10 = ["same"]
        self.lambda12 = ["same"]
        self.lambda13 = ["same"]
        self.lambda14 = ["same"]
        self.lambda20 = ["same"]
        self.lambda22 = ["same"]
        self.lambda23 = ["same"]
        self.lambda24 = ["same"]
        self.lambda30 = ["same"]
        self.lambda32 = ["same"]
        self.lambda33 = ["same"]
        self.lambda34 = ["same"]
        self.r_b_bar = ["same"]
        self.P_bL_bar = ["same"]

        #Computed flows.
        self.r_b = [r_b_1]
        self.P_bL = [P_bL_1]
        self.r_bL = ["same"]
        self.ERr_bL = ["same"]
        self.C = ["same"]
        self.Y = ["same"]
        self.T = ["same"]
        self.YDr = [YDr_1]
        self.YDr_e = [YDr_1]
        self.CG = ["same"]
        self.V = [V_1]
        self.Ve = ["same"]
        self.Hd = ["same"]
        self.Bh = [Bh_1]
        self.BLh = [BLh_1]
        self.Hh = ["same"]
        self.dBs = ["same"]
        self.Bs = [Bs_1]
        self.Bcb = [Bcb_1]
        self.dHs = ["same"]
        self.Hs = [Hs_1]

        #One explanation list per calculated variable, for printExplanations().
        self.r_b_expl = [""]
        self.P_bL_expl = [""]
        self.r_bL_expl = [""]
        self.ERr_bL_expl = [""]
        self.C_expl = [""]
        self.Y_expl = [""]
        self.T_expl = [""]
        self.YDr_expl = [""]
        self.CG_expl = [""]
        self.V_expl = [""]
        self.Ve_expl = [""]
        self.Hd_expl = [""]
        self.Bh_expl = [""]
        self.BLh_expl = [""]
        self.Hh_expl = [""]
        self.dBs_expl = [""]
        self.Bs_expl = [""]
        self.Bcb_expl = [""]
        self.dHs_expl = [""]
        self.Hs_expl = [""]

    def get_V(self):
        return self.V[-1]

    def get_YDr(self):
        return self.YDr[-1]

    def get_Bh(self):
        return self.Bh[-1]

    def get_BLh(self):
        return self.BLh[-1]

    def get_Bs(self):
        return self.Bs[-1]

    def get_Bcb(self):
        return self.Bcb[-1]

    def get_Hs(self):
        return self.Hs[-1]

    def get_r_b(self):
        return self.r_b[-1]

    def get_P_bL(self):
        return self.P_bL[-1]

    def set_G(self, v):
        self.G.append(v)

    def set_alpha1(self, v):
        self.alpha1.append(v)

    def set_alpha2(self, v):
        self.alpha2.append(v)

    def set_theta(self, v):
        self.theta.append(v)

    def set_lambda10(self, v):
        self.lambda10.append(v)

    def set_lambda12(self, v):
        self.lambda12.append(v)

    def set_lambda13(self, v):
        self.lambda13.append(v)

    def set_lambda14(self, v):
        self.lambda14.append(v)

    def set_lambda20(self, v):
        self.lambda20.append(v)

    def set_lambda22(self, v):
        self.lambda22.append(v)

    def set_lambda23(self, v):
        self.lambda23.append(v)

    def set_lambda24(self, v):
        self.lambda24.append(v)

    def set_lambda30(self, v):
        self.lambda30.append(v)

    def set_lambda32(self, v):
        self.lambda32.append(v)

    def set_lambda33(self, v):
        self.lambda33.append(v)

    def set_lambda34(self, v):
        self.lambda34.append(v)

    def set_r_b_bar(self, v):
        self.r_b_bar.append(v)

    def set_P_bL_bar(self, v):
        self.P_bL_bar.append(v)

    def set_r_b(self, v):
        self.r_b.append(v)

    def set_P_bL(self, v):
        self.P_bL.append(v)

    def set_r_bL(self, v):
        self.r_bL.append(v)

    def set_ERr_bL(self, v):
        self.ERr_bL.append(v)

    def set_C(self, v):
        self.C.append(v)

    def set_Y(self, v):
        self.Y.append(v)

    def set_T(self, v):
        self.T.append(v)

    def set_YDr(self, v):
        self.YDr.append(v)

    def set_YDr_e(self, v):
        self.YDr_e.append(v)

    def set_CG(self, v):
        self.CG.append(v)

    def set_V(self, v):
        self.V.append(v)

    def set_Ve(self, v):
        self.Ve.append(v)

    def set_Hd(self, v):
        self.Hd.append(v)

    def set_Bh(self, v):
        self.Bh.append(v)

    def set_BLh(self, v):
        self.BLh.append(v)

    def set_Hh(self, v):
        self.Hh.append(v)

    def set_dBs(self, v):
        self.dBs.append(v)

    def set_Bs(self, v):
        self.Bs.append(v)

    def set_Bcb(self, v):
        self.Bcb.append(v)

    def set_dHs(self, v):
        self.dHs.append(v)

    def set_Hs(self, v):
        self.Hs.append(v)

    def progresCycle(self, G, alpha1, alpha2, theta, lambda10, lambda12, lambda13,
                      lambda14, lambda20, lambda22, lambda23, lambda24, lambda30,
                      lambda32, lambda33, lambda34, r_b_bar, P_bL_bar):
        period_number = self.period[-1] + 1

        V_1 = self.get_V()
        YDr_1 = self.get_YDr()
        Bh_1 = self.get_Bh()
        BLh_1 = self.get_BLh()
        Bs_1 = self.get_Bs()
        Bcb_1 = self.get_Bcb()
        Hs_1 = self.get_Hs()
        r_1 = self.get_r_b()
        PbL_1 = self.get_P_bL()

        #Exogenous rates and the long bond price
        r_b, r_b_expl = identity(r_b_bar, "The exogenous bill rate (r_b_bar)",
                                  "The bill rate (r_b)", "r_b = r_b_bar", period_number)
        P_bL, P_bL_expl = identity(P_bL_bar, "The exogenous long bond price peg (P_bL_bar)",
                                    "The long bond price (P_bL)", "P_bL = P_bL_bar", period_number)
        r_bL, r_bL_expl = div("same", P_bL,
                               "The fixed coupon payment (a constant, always the same)",
                               "The long bond price",
                               "The long bond yield (r_bL)", "r_bL = 1 / P_bL", period_number)
        ERr_bL, ERr_bL_expl = identity(r_bL, "The long bond yield (r_bL)",
                                        "The expected return on long bonds (ERr_bL)",
                                        "ERr_bL = r_bL, since bond-price expectations are "
                                        "static (p_bL^e = P_bL) in this model", period_number)

        #Households: expected income and consumption
        YDr_e = YDr_1  # expected regular disposable income = last period's actual
        income_effect, income_effect_expl = mult(alpha1, YDr_e,
                                                   "The propensity to consume out of income (alpha1)",
                                                   "Expected regular disposable income",
                                                   "the income effect on consumption",
                                                   "alpha1 * YD_r^e", period_number)
        wealth_effect, wealth_effect_expl = mult(alpha2, V_1,
                                                  "The propensity to consume out of wealth (alpha2)",
                                                  "Previous wealth",
                                                  "the wealth effect on consumption",
                                                  "alpha2 * V(-1)", period_number)
        C, C_expl_add = add(income_effect, wealth_effect,
                             "The income effect on consumption",
                             "The wealth effect on consumption",
                             "Consumption (C)", "C = alpha1*YD_r^e + alpha2*V(-1)", period_number)
        C_expl = income_effect_expl + "\n" + wealth_effect_expl + "\n" + C_expl_add

        Y, Y_expl = add(C, G, "Consumption", "Government spending",
                         "Income (Y)", "Y = C + G", period_number)

        #Taxes and regular disposable income
        interest_bills, interest_bills_expl = mult(r_1, Bh_1,
                                                     "The (lagged) bill rate",
                                                     "The previous stock of bills held",
                                                     "Interest income on bills",
                                                     "r(-1) * B_h(-1)", period_number)
        property_income, property_income_expl = add(interest_bills, BLh_1,
                                                      "Interest income on bills",
                                                      "Coupon income on long bonds held",
                                                      "Total property income",
                                                      "r(-1)*B_h(-1) + BL_h(-1)", period_number)
        tax_base, tax_base_expl = add(Y, property_income, "Income", "Total property income",
                                       "The tax base",
                                       "Y + r(-1)*B_h(-1) + BL_h(-1)", period_number)
        T, T_expl_mult = mult(theta, tax_base, "The tax rate (theta)", "The tax base",
                               "Taxes (T)",
                               "T = theta*(Y + r(-1)*B_h(-1) + BL_h(-1))", period_number)
        T_expl = (interest_bills_expl + "\n" + property_income_expl + "\n"
                  + tax_base_expl + "\n" + T_expl_mult)

        income_after_tax, income_after_tax_expl = sub(Y, T, "Income", "Taxes",
                                                        "Income after tax", "Y - T", period_number)
        YDr, YDr_expl_add = add(income_after_tax, property_income,
                                 "Income after tax", "Total property income",
                                 "Regular disposable income (YD_r)",
                                 "YD_r = Y - T + r(-1)*B_h(-1) + BL_h(-1)", period_number)
        YDr_expl = income_after_tax_expl + "\n" + YDr_expl_add

        #Wealth and capital gains
        delta_PbL, delta_PbL_expl = sub(P_bL, PbL_1, "The current long bond price",
                                         "The previous long bond price",
                                         "The change in the long bond price",
                                         "delta_P_bL = P_bL - P_bL(-1)", period_number)
        CG, CG_expl_mult = mult(delta_PbL, BLh_1, "The change in the long bond price",
                                 "The previous stock of long bonds held",
                                 "Capital gains (CG)", "CG = delta_P_bL * BL_h(-1)", period_number)
        CG_expl = delta_PbL_expl + "\n" + CG_expl_mult

        delta_V, delta_V_expl = sub(YDr, C, "Regular disposable income", "Consumption",
                                     "The change in wealth from saving",
                                     "YD_r - C", period_number)
        V_partial, V_partial_expl = add(V_1, delta_V, "Previous wealth",
                                         "The change in wealth from saving",
                                         "Wealth before capital gains",
                                         "V(-1) + (YD_r - C)", period_number)
        V, V_expl_add = add(V_partial, CG, "Wealth before capital gains", "Capital gains",
                             "Wealth (V)", "V = V(-1) + (YD_r - C) + CG", period_number)
        V_expl = delta_V_expl + "\n" + V_partial_expl + "\n" + V_expl_add

        delta_Ve, delta_Ve_expl = sub(YDr_e, C, "Expected regular disposable income",
                                       "Consumption", "The expected change in wealth from saving",
                                       "YD_r^e - C", period_number)
        Ve_partial, Ve_partial_expl = add(V_1, delta_Ve, "Previous wealth",
                                           "The expected change in wealth from saving",
                                           "Expected wealth before capital gains",
                                           "V(-1) + (YD_r^e - C)", period_number)
        Ve, Ve_expl_add = add(Ve_partial, CG, "Expected wealth before capital gains",
                               "Capital gains", "Expected wealth (V^e)",
                               "V^e = V(-1) + (YD_r^e - C) + CG", period_number)
        Ve_expl = delta_Ve_expl + "\n" + Ve_partial_expl + "\n" + Ve_expl_add

        #Portfolio allocation across money, bills and long bonds
        ratio, ratio_expl = div(YDr_e, Ve, "Expected regular disposable income",
                                 "Expected wealth", "The expected income-to-wealth ratio",
                                 "YD_r^e / V^e", period_number)

        term_r_h, term_r_h_expl = mult(lambda12, r_b, "lambda12", "The bill rate",
                                        "The bill-rate effect on money demand",
                                        "lambda12 * r_b", period_number)
        term_ER_h, term_ER_h_expl = mult(lambda13, ERr_bL, "lambda13",
                                          "The expected return on long bonds",
                                          "The bond-return effect on money demand",
                                          "lambda13 * ERr_bL", period_number)
        term_ratio_h, term_ratio_h_expl = mult(lambda14, ratio, "lambda14",
                                                "The expected income-to-wealth ratio",
                                                "The income effect on money demand",
                                                "lambda14 * (YD_r^e/V^e)", period_number)
        share_h1, share_h1_expl = add(lambda10, term_r_h, "lambda10",
                                       "The bill-rate effect on money demand",
                                       "Money-demand share, step 1",
                                       "lambda10 + lambda12*r_b", period_number)
        share_h2, share_h2_expl = add(share_h1, term_ER_h, "Money-demand share, step 1",
                                       "The bond-return effect on money demand",
                                       "Money-demand share, step 2",
                                       "+ lambda13*ERr_bL", period_number)
        Hd_share, Hd_share_expl = add(share_h2, term_ratio_h, "Money-demand share, step 2",
                                       "The income effect on money demand",
                                       "The desired money share of wealth (H^d/V^e)",
                                       "H^d/V^e = lambda10+lambda12*r_b+lambda13*ERr_bL"
                                       "+lambda14*(YD_r^e/V^e)", period_number)
        Hd, Hd_expl_mult = mult(Hd_share, Ve, "The desired money share of wealth",
                                 "Expected wealth", "Desired money holdings (H^d)",
                                 "H^d = (H^d/V^e) * V^e", period_number)
        Hd_expl = (ratio_expl + "\n" + term_r_h_expl + "\n" + term_ER_h_expl + "\n"
                   + term_ratio_h_expl + "\n" + share_h1_expl + "\n" + share_h2_expl + "\n"
                   + Hd_share_expl + "\n" + Hd_expl_mult)

        term_r_b, term_r_b_expl = mult(lambda22, r_b, "lambda22", "The bill rate",
                                        "The bill-rate effect on bill demand",
                                        "lambda22 * r_b", period_number)
        term_ER_b, term_ER_b_expl = mult(lambda23, ERr_bL, "lambda23",
                                          "The expected return on long bonds",
                                          "The bond-return effect on bill demand",
                                          "lambda23 * ERr_bL", period_number)
        term_ratio_b, term_ratio_b_expl = mult(lambda24, ratio, "lambda24",
                                                "The expected income-to-wealth ratio",
                                                "The income effect on bill demand",
                                                "lambda24 * (YD_r^e/V^e)", period_number)
        share_b1, share_b1_expl = add(lambda20, term_r_b, "lambda20",
                                       "The bill-rate effect on bill demand",
                                       "Bill-demand share, step 1",
                                       "lambda20 + lambda22*r_b", period_number)
        share_b2, share_b2_expl = add(share_b1, term_ER_b, "Bill-demand share, step 1",
                                       "The bond-return effect on bill demand",
                                       "Bill-demand share, step 2",
                                       "+ lambda23*ERr_bL", period_number)
        Bh_share, Bh_share_expl = add(share_b2, term_ratio_b, "Bill-demand share, step 2",
                                       "The income effect on bill demand",
                                       "The desired bill share of wealth (B^d/V^e)",
                                       "B^d/V^e = lambda20+lambda22*r_b+lambda23*ERr_bL"
                                       "+lambda24*(YD_r^e/V^e)", period_number)
        Bh, Bh_expl_mult = mult(Bh_share, Ve, "The desired bill share of wealth",
                                 "Expected wealth",
                                 "Desired and actual bill holdings (B_h), where B_h = B^d",
                                 "B_h = B^d = (B^d/V^e) * V^e", period_number)
        Bh_expl = (term_r_b_expl + "\n" + term_ER_b_expl + "\n" + term_ratio_b_expl + "\n"
                   + share_b1_expl + "\n" + share_b2_expl + "\n" + Bh_share_expl + "\n"
                   + Bh_expl_mult)

        term_r_bl, term_r_bl_expl = mult(lambda32, r_b, "lambda32", "The bill rate",
                                          "The bill-rate effect on long bond demand",
                                          "lambda32 * r_b", period_number)
        term_ER_bl, term_ER_bl_expl = mult(lambda33, ERr_bL, "lambda33",
                                            "The expected return on long bonds",
                                            "The bond-return effect on long bond demand",
                                            "lambda33 * ERr_bL", period_number)
        term_ratio_bl, term_ratio_bl_expl = mult(lambda34, ratio, "lambda34",
                                                  "The expected income-to-wealth ratio",
                                                  "The income effect on long bond demand",
                                                  "lambda34 * (YD_r^e/V^e)", period_number)
        share_bl1, share_bl1_expl = add(lambda30, term_r_bl, "lambda30",
                                         "The bill-rate effect on long bond demand",
                                         "Long-bond-demand share, step 1",
                                         "lambda30 + lambda32*r_b", period_number)
        share_bl2, share_bl2_expl = add(share_bl1, term_ER_bl, "Long-bond-demand share, step 1",
                                         "The bond-return effect on long bond demand",
                                         "Long-bond-demand share, step 2",
                                         "+ lambda33*ERr_bL", period_number)
        BLh_share, BLh_share_expl = add(share_bl2, term_ratio_bl, "Long-bond-demand share, step 2",
                                         "The income effect on long bond demand",
                                         "The desired long bond value share of wealth "
                                         "(BL^d*P_bL/V^e)",
                                         "BL^d*P_bL/V^e = lambda30+lambda32*r_b+lambda33*ERr_bL"
                                         "+lambda34*(YD_r^e/V^e)", period_number)
        BLh_value, BLh_value_expl = mult(BLh_share, Ve, "The desired long bond value share of wealth",
                                          "Expected wealth",
                                          "The desired value of long bond holdings (BL^d * P_bL)",
                                          "BL^d*P_bL = (BL^d*P_bL/V^e) * V^e", period_number)
        BLh, BLh_expl_div = div(BLh_value, P_bL, "The desired value of long bond holdings",
                                 "The long bond price",
                                 "Desired and actual long bond holdings (BL_h), "
                                 "where BL_h = BL^d = BL_s",
                                 "BL_h = (BL^d*P_bL) / P_bL", period_number)
        BLh_expl = (term_r_bl_expl + "\n" + term_ER_bl_expl + "\n" + term_ratio_bl_expl + "\n"
                    + share_bl1_expl + "\n" + share_bl2_expl + "\n" + BLh_share_expl + "\n"
                    + BLh_value_expl + "\n" + BLh_expl_div)

        #Realized money holdings
        term_PbL_BLh, term_PbL_BLh_expl = mult(P_bL, BLh, "The long bond price",
                                                "Long bond holdings",
                                                "The value of long bond holdings",
                                                "P_bL * BL_h", period_number)
        step_hh1, step_hh1_expl = sub(V, Bh, "Wealth", "Bill holdings", "Wealth net of bills",
                                       "V - B_h", period_number)
        Hh, Hh_expl_sub = sub(step_hh1, term_PbL_BLh, "Wealth net of bills",
                               "The value of long bond holdings",
                               "Realized money holdings (H_h)",
                               "H_h = V - B_h - P_bL*BL_h", period_number)
        Hh_expl = term_PbL_BLh_expl + "\n" + step_hh1_expl + "\n" + Hh_expl_sub

        #Government financing
        govt_interest, govt_interest_expl = mult(r_1, Bs_1, "The (lagged) bill rate",
                                                  "The previous stock of bills supplied",
                                                  "Government interest payments on bills",
                                                  "r(-1)*B_s(-1)", period_number)
        govt_outlays_partial, govt_outlays_partial_expl = add(G, govt_interest,
                                                                "Government spending",
                                                                "Government interest payments on bills",
                                                                "Government outlays before bond coupons",
                                                                "G + r(-1)*B_s(-1)", period_number)
        govt_outlays, govt_outlays_expl_add = add(govt_outlays_partial, BLh_1,
                                                   "Government outlays before bond coupons",
                                                   "Coupon payments on long bonds outstanding",
                                                   "Total government outlays",
                                                   "G + r(-1)*B_s(-1) + BL_s(-1)", period_number)
        govt_outlays_expl = (govt_interest_expl + "\n" + govt_outlays_partial_expl
                              + "\n" + govt_outlays_expl_add)

        cb_interest, cb_interest_expl = mult(r_1, Bcb_1, "The (lagged) bill rate",
                                              "The previous stock of bills held by the central bank",
                                              "Interest received by the central bank",
                                              "r(-1)*B_cb(-1)", period_number)
        govt_receipts, govt_receipts_expl = add(T, cb_interest, "Taxes",
                                                 "Interest remitted by the central bank",
                                                 "Total government receipts",
                                                 "T + r(-1)*B_cb(-1)", period_number)

        deficit, deficit_expl = sub(govt_outlays, govt_receipts, "Total government outlays",
                                     "Total government receipts",
                                     "The government's financing need",
                                     "outlays - receipts", period_number)

        delta_BLs, delta_BLs_expl = sub(BLh, BLh_1, "Long bond holdings this period",
                                         "Long bond holdings last period",
                                         "The change in long bonds supplied",
                                         "delta_BL_s = BL_h - BL_h(-1), since BL_s = BL_h",
                                         period_number)
        BLs_proceeds, BLs_proceeds_expl = mult(delta_BLs, P_bL, "The change in long bonds supplied",
                                                "The long bond price",
                                                "Proceeds from new long bond issuance",
                                                "delta_BL_s * P_bL", period_number)
        dBs, dBs_expl_sub = sub(deficit, BLs_proceeds, "The government's financing need",
                                 "Proceeds from new long bond issuance",
                                 "The change in bills supplied (delta_B_s)",
                                 "delta_B_s = financing need - delta_BL_s*P_bL", period_number)
        dBs_expl = (govt_outlays_expl + "\n" + cb_interest_expl + "\n" + govt_receipts_expl
                    + "\n" + deficit_expl + "\n" + delta_BLs_expl + "\n" + BLs_proceeds_expl
                    + "\n" + dBs_expl_sub)

        Bs, Bs_expl_add = add(Bs_1, dBs, "The previous stock of bills supplied",
                               "The change in bills supplied", "Bills supplied (B_s)",
                               "B_s = B_s(-1) + delta_B_s", period_number)
        Bs_expl = dBs_expl + "\n" + Bs_expl_add

        #Central bank and the money supply
        Bcb, Bcb_expl = sub(Bs, Bh, "Bills supplied", "Bills held by households",
                             "Central bank bill holdings (B_cb)", "B_cb = B_s - B_h",
                             period_number)

        delta_Bcb, delta_Bcb_expl = sub(Bcb, Bcb_1, "Central bank bill holdings this period",
                                         "Central bank bill holdings last period",
                                         "The change in central bank bill holdings",
                                         "B_cb - B_cb(-1)", period_number)
        dHs, dHs_expl_id = identity(delta_Bcb, "The change in central bank bill holdings",
                                     "The change in money supplied (delta_H_s)",
                                     "delta_H_s = delta_B_cb", period_number)
        dHs_expl = delta_Bcb_expl + "\n" + dHs_expl_id

        Hs, Hs_expl_add = add(Hs_1, dHs, "The previous stock of money supplied",
                               "The change in money supplied",
                               "Money supplied (H_s), which should equal H_h "
                               "(the hidden equation)",
                               "H_s = H_s(-1) + delta_H_s", period_number)
        Hs_expl = dHs_expl + "\n" + Hs_expl_add

        self.period.append(period_number)
        self.set_G(G)
        self.set_alpha1(alpha1)
        self.set_alpha2(alpha2)
        self.set_theta(theta)
        self.set_lambda10(lambda10)
        self.set_lambda12(lambda12)
        self.set_lambda13(lambda13)
        self.set_lambda14(lambda14)
        self.set_lambda20(lambda20)
        self.set_lambda22(lambda22)
        self.set_lambda23(lambda23)
        self.set_lambda24(lambda24)
        self.set_lambda30(lambda30)
        self.set_lambda32(lambda32)
        self.set_lambda33(lambda33)
        self.set_lambda34(lambda34)
        self.set_r_b_bar(r_b_bar)
        self.set_P_bL_bar(P_bL_bar)

        self.set_r_b(r_b)
        self.set_P_bL(P_bL)
        self.set_r_bL(r_bL)
        self.set_ERr_bL(ERr_bL)
        self.set_C(C)
        self.set_Y(Y)
        self.set_T(T)
        self.set_YDr(YDr)
        self.set_YDr_e(YDr_e)
        self.set_CG(CG)
        self.set_V(V)
        self.set_Ve(Ve)
        self.set_Hd(Hd)
        self.set_Bh(Bh)
        self.set_BLh(BLh)
        self.set_Hh(Hh)
        self.set_dBs(dBs)
        self.set_Bs(Bs)
        self.set_Bcb(Bcb)
        self.set_dHs(dHs)
        self.set_Hs(Hs)

        self.r_b_expl.append(r_b_expl)
        self.P_bL_expl.append(P_bL_expl)
        self.r_bL_expl.append(r_bL_expl)
        self.ERr_bL_expl.append(ERr_bL_expl)
        self.C_expl.append(C_expl)
        self.Y_expl.append(Y_expl)
        self.T_expl.append(T_expl)
        self.YDr_expl.append(YDr_expl)
        self.CG_expl.append(CG_expl)
        self.V_expl.append(V_expl)
        self.Ve_expl.append(Ve_expl)
        self.Hd_expl.append(Hd_expl)
        self.Bh_expl.append(Bh_expl)
        self.BLh_expl.append(BLh_expl)
        self.Hh_expl.append(Hh_expl)
        self.dBs_expl.append(dBs_expl)
        self.Bs_expl.append(Bs_expl)
        self.Bcb_expl.append(Bcb_expl)
        self.dHs_expl.append(dHs_expl)
        self.Hs_expl.append(Hs_expl)

    def printHistory(self):
        #Variables down the rows, periods across the columns (same layout as Chapter 3).
        variables = [
            ("G", self.G), ("a1", self.alpha1), ("a2", self.alpha2), ("theta", self.theta),
            ("l10", self.lambda10), ("l12", self.lambda12), ("l13", self.lambda13),
            ("l14", self.lambda14), ("l20", self.lambda20), ("l22", self.lambda22),
            ("l23", self.lambda23), ("l24", self.lambda24), ("l30", self.lambda30),
            ("l32", self.lambda32), ("l33", self.lambda33), ("l34", self.lambda34),
            ("rb_bar", self.r_b_bar), ("PbL_bar", self.P_bL_bar),
            ("rb", self.r_b), ("PbL", self.P_bL), ("rbL", self.r_bL), ("ERrbL", self.ERr_bL),
            ("C", self.C), ("Y", self.Y), ("T", self.T), ("YDr", self.YDr),
            ("YDr^e", self.YDr_e), ("CG", self.CG), ("V", self.V), ("V^e", self.Ve),
            ("Hd", self.Hd), ("Bh", self.Bh), ("BLh", self.BLh), ("Hh", self.Hh),
            ("dBs", self.dBs), ("Bs", self.Bs), ("Bcb", self.Bcb), ("dHs", self.dHs),
            ("Hs", self.Hs),
        ]

        label_width = 8
        col_width = 10

        header_line = f"{'Variable':<{label_width}}" + "".join(
            f"{'Period ' + str(p):>{col_width}}" for p in self.period)
        total_width = len(header_line)

        print("=" * total_width)
        print("SIMULATION RESULTS \u2014 Model LP")
        print("=" * total_width)
        print(header_line)
        print("-" * total_width)
        for name, values in variables:
            row_line = f"{name:<{label_width}}" + "".join(
                f"{str(v):>{col_width}}" for v in values)
            print(row_line)
        print("=" * total_width)

        #Ratio of unknown / same / up / down answers across the table
        #Counts every value in every variable's history, excluding each
        #variable's period-0 initial condition (that is a starting point,
        #not a computed answer). The numbers are unchanged from before
        #only how they're displayed (via _print_answer_distribution) has
        #changed.
        counts = {"up": 0, "down": 0, "same": 0, "unknown": 0}
        for name, values in variables:
            for v in values[1:]:
                if v in counts:
                    counts[v] += 1
        total = sum(counts.values())
        if total > 0:
            self._print_answer_distribution(counts, total)

    def _print_answer_distribution(self, counts, total):
        #A small, purely cosmetic "dashboard" box summarising how the
        #up/down/same/unknown answers were distributed across every
        #computed value in the table above. The counts themselves come
        #straight from printHistory() - this function only formats them.
        box_width = 64
        inner = box_width - 2

        def boxed(text=""):
            return "\u2502" + text.ljust(inner) + "\u2502"

        top = "\u256d" + "\u2500" * inner + "\u256e"
        divider = "\u251c" + "\u2500" * inner + "\u2524"
        bottom = "\u2570" + "\u2500" * inner + "\u256f"

        print()
        print(top)
        print(boxed(" ANSWER DISTRIBUTION".center(inner)))
        print(divider)
        print(boxed(f"  Unknown : Same : Up : Down  =  {counts['unknown']} : "
                     f"{counts['same']} : {counts['up']} : {counts['down']}"
                     f"   (n = {total})"))
        print(boxed())

        bar_slots = 24
        for key in ("unknown", "same", "up", "down"):
            pct = counts[key] / total
            filled = round(pct * bar_slots)
            bar = "\u2588" * filled + "\u2591" * (bar_slots - filled)
            row = f"  {key.capitalize():<8}{bar}  {pct:>5.1%}"
            print(boxed(row))

        print(bottom)

    def printExplanations(self):
        #Full reasoning trace per period: one underlined heading + bullets per variable.
        width = 60
        for i in range(1, len(self.period)):
            print()
            print("=" * width)
            print(f"PERIOD {self.period[i]} \u2014 REASONING TRACE".center(width))
            print("=" * width)
            sections = [
                ("Bill rate (r_b)", self.r_b_expl[i]),
                ("Long bond price (P_bL)", self.P_bL_expl[i]),
                ("Long bond yield (r_bL)", self.r_bL_expl[i]),
                ("Expected return on long bonds (ERr_bL)", self.ERr_bL_expl[i]),
                ("Consumption (C)", self.C_expl[i]),
                ("Income (Y)", self.Y_expl[i]),
                ("Taxes (T)", self.T_expl[i]),
                ("Regular disposable income (YD_r)", self.YDr_expl[i]),
                ("Capital gains (CG)", self.CG_expl[i]),
                ("Wealth (V)", self.V_expl[i]),
                ("Expected wealth (V^e)", self.Ve_expl[i]),
                ("Desired money holdings (H^d)", self.Hd_expl[i]),
                ("Bill holdings (B_h)", self.Bh_expl[i]),
                ("Long bond holdings (BL_h)", self.BLh_expl[i]),
                ("Realized money holdings (H_h)", self.Hh_expl[i]),
                ("Change in bills supplied (delta_B_s)", self.dBs_expl[i]),
                ("Bills supplied (B_s)", self.Bs_expl[i]),
                ("Central bank bill holdings (B_cb)", self.Bcb_expl[i]),
                ("Change in money supplied (delta_H_s)", self.dHs_expl[i]),
                ("Money supplied (H_s)", self.Hs_expl[i]),
            ]
            for title, text in sections:
                print()
                print(title)
                print("\u2500" * len(title))
                for line_text in text.split("\n"):
                    if line_text.strip():
                        print(f"  \u2022 {line_text}")
        print()
        print("=" * width)


qecon = QualitativeEconomyLP(V_1="same", YDr_1="same", Bh_1="same", BLh_1="same",
                              Bs_1="same", Bcb_1="same", Hs_1="same",
                              r_b_1="same", P_bL_1="same")
#Period 1: a rise in government spending (G up), everything else same.
qecon.progresCycle(G="up", alpha1="same", alpha2="same", theta="same",
                    lambda10="same", lambda12="same", lambda13="same", lambda14="same",
                    lambda20="same", lambda22="same", lambda23="same", lambda24="same",
                    lambda30="same", lambda32="same", lambda33="same", lambda34="same",
                    r_b_bar="same", P_bL_bar="same")
#Period 2: no further changes, everything else same.
qecon.progresCycle(G="same", alpha1="same", alpha2="same", theta="same",
                    lambda10="same", lambda12="same", lambda13="same", lambda14="same",
                    lambda20="same", lambda22="same", lambda23="same", lambda24="same",
                    lambda30="same", lambda32="same", lambda33="same", lambda34="same",
                    r_b_bar="same", P_bL_bar="same")
qecon.printHistory()
qecon.printExplanations()
```
